```python
df = pd.read_csv('Customer Segmentation.csv')
df
```

```python
sns.heatmap(df.isnull(),
            yticklabels=False,
            cbar=False,
            cmap='viridis')
```

```python
df.nunique()
```

```python
plt.figure(figsize=(12, 6))
department = df['department']
ax = sns.countplot(y=department, data=df, orient='h', order=department.value_counts().index)
plt.title('setores', fontsize=18)
ax.bar_label(ax.containers[0])
plt.xlabel('Quantidade', fontsize=16)
plt.ylabel('departamento', fontsize=16)
```

```python
plt.figure(figsize=(12, 6))
sex = df['gender']
ax = sns.countplot(x=sex, data=df, orient='v', order=sex.value_counts().sort_values().index)
plt.title('Distribuição de genero', fontsize=18)
ax.bar_label(ax.containers[0])
plt.ylabel('Quantidade', fontsize=16)
plt.xlabel('Genero', fontsize=16)
```

```python
df['time_zone'] = df['time_zone'].str.split('/').str[0]
plt.figure(figsize=(12, 6))
continent = df['time_zone']
ax=sns.countplot(x='time_zone' ,data =df, orient='v', order=continent.value_counts().sort_values().index)
plt.title('Região', fontsize=18)
ax.bar_label(ax.containers[0])
plt.ylabel('Quantidade', fontsize=16)
plt.xlabel('Região', fontsize=16)
```

```python
plt.figure(figsize=(20, 10))
sizes = df['country'].value_counts()[:40]
labels = sizes.index.tolist()
squarify.plot(sizes=sizes, label=labels, alpha=0.6)
plt.title('Top 40 Países', fontsize=18)
plt.axis('off')
plt.show()
```
